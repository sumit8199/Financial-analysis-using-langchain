```python
import pandas as pd

data = pd.read_excel("/content/My_Financial_Report_Dataset.xlsx")
```

```python
data.head()
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
kpi_prompt = PromptTemplate(
    input_veriables = ["data"],
    template = """
You are a financial analyst and you have been provided with a financial data of a company.
Extract the most important financial and buisness KPI's from the data provided.
Ensure you give me top 10 most important kpis only and present them in a bullet point format.
financial_data:{data}""")

print(kpi_prompt.format(data = data))
```

```python
chain_1 = kpi_prompt | llm | StrOutputParser()
```

```python
kpis = chain_1.invoke({"data":data})
```

```python
print(kpis)
```

```python
benchmark_prompt = PromptTemplate(
    input_veriables=[kpis],
    template=""" You are Senior Strategy Consultant and your job is to critically analyze the KPIs provided
    and establish appropriate performance benchmark and management evaluation.
    From the top 10 KPIs provided, Ensure that you identify the most important KPIs and for them provide the following:
    -Define a reasonable target/benchmark
    -state and carrent performance of the company.
    -classify the performance of this company:
      -Strong
      -Moderate
      -Weak or needs improvement

    Present the output in form of Structured Table """)
```

```python
chain_2 = benchmark_prompt | llm | StrOutputParser()
```

```python
benchmark = chain_2.invoke({"kpis":kpis})
print(benchmark)
```

```python
report_prompt = PromptTemplate(
    input_variables=["benchmark"],
    template="""You are a Chief Financial Officer (CFO).
Using the benchmark evaluation and performance analysis provided below, draft an official, comprehensive Executive Report for the Board of Directors.

The report should include:
1. Executive Summary: High-level overview of the financial health of the business.
2. Key Operational Strengths: Analysis of where the company is performing strongly.
3. Risk Factors & Vulnerabilities: Critique of areas needing immediate management action and intervention.
4. Strategic Recommendations: Practical steps for the upcoming quarters to drive sustainable growth.

Ensure the tone is highly professional, strategic, and concise.

Benchmark Evaluation:
{benchmark}"""
)

chain_3 = report_prompt | llm | StrOutputParser()
executive_report = chain_3.invoke({"benchmark": benchmark})
print(executive_report)
```
